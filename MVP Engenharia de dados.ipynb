
# 1. Contexto de Negócios e Perguntas
## 1.1 Contexto

A indústria de videogames é composta por dois elementos diretamente relacionados: o desenvolvimento e comercialização de jogos e a evolução das plataformas de hardware utilizadas para executá-los. Ao longo das diferentes gerações de consoles, novos fabricantes e plataformas surgiram, enquanto outras foram descontinuadas, modificando o cenário do mercado de jogos.

Este MVP tem como foco analisar a evolução dos consoles domésticos e o desempenho comercial dos jogos lançados para essas plataformas no período de 1995 a 2020. Para isso, serão combinados dois conjuntos de dados públicos disponibilizados no Kaggle: o dataset Console_Data, que contém informações sobre consoles domésticos, fabricantes, gerações, períodos de comercialização e unidades vendidas, e o dataset Video Games Sales, que contém informações sobre jogos, plataformas, gêneros, publishers, developers, avaliações de críticos e vendas por região.

A utilização conjunta dessas fontes permite analisar o mercado sob duas perspectivas complementares. A primeira está relacionada ao hardware, considerando a evolução das gerações de consoles e o volume de unidades comercializadas. A segunda está relacionada ao software, considerando os jogos lançados, seus gêneros, vendas globais e regionais e avaliações dos críticos.

O período de análise integrada será definido entre 1995 e 2020, pois corresponde ao intervalo de cobertura temporal comum aos dois datasets. Os anos anteriores a 1995 presentes no dataset de consoles poderão ser mantidos na camada de dados, mas não serão utilizados nas análises que dependem da integração entre as duas fontes.

## 1.2 Problema

A análise isolada de vendas de consoles ou de vendas de jogos apresenta apenas uma visão parcial do mercado. Para compreender melhor a evolução das plataformas, é necessário relacionar as informações de hardware com os dados dos jogos comercializados nessas plataformas.

O problema abordado neste MVP é:

Como a evolução das plataformas de consoles domésticos está relacionada ao desempenho comercial dos jogos entre 1995 e 2020?

Para investigar esse problema, será necessário integrar informações provenientes das duas fontes, padronizando a identificação das plataformas e estabelecendo uma relação entre os consoles presentes no Console_Data e as plataformas presentes no Video Games Sales.

A integração permitirá analisar diferentes aspectos do mercado, como gerações de consoles, fabricantes, gêneros de jogos, vendas regionais e avaliações de críticos.

## 1.3 Objetivo geral

Construir um pipeline de dados em nuvem capaz de integrar, transformar, armazenar e analisar dados sobre consoles domésticos e vendas de jogos, permitindo investigar a evolução das plataformas e sua relação com o desempenho comercial dos jogos entre 1995 e 2020.

## 1.6 Escopo e limitações iniciais

Fonte: Kaggle – Video Game Sales and Industry Data 1980-2024

- `critic_score`: avaliação dos críticos.

- `na_sales`: vendas na América do Norte em milhões.
- `jp_sales`: vendas no Japão em milhões.
- `pal_sales`: vendas na Europa e África em milhões.
- `other_sales`: vendas nas demais regiões em milhões.

# 3. Modelagem e Catálogo de Dados

A etapa de modelagem organiza os dados carregados na etapa anterior de forma a facilitar sua transformação, integração e posterior utilização nas análises. A modelagem foi definida considerando as perguntas de negócio apresentadas na Seção 1 e as características das duas fontes utilizadas no projeto.

Neste MVP, foi adotada uma abordagem baseada em **Esquema Estrela (Star Schema)** para a camada de consumo analítico, mantendo inicialmente os dados tratados em tabelas Silver próximas à estrutura das fontes originais.

A escolha do Esquema Estrela permite separar informações descritivas, como consoles, plataformas e jogos, das métricas utilizadas nas análises, como unidades vendidas e vendas regionais. Essa organização também facilita consultas relacionadas às gerações de consoles, plataformas, gêneros, regiões e avaliações dos críticos.

## 3.1 Arquitetura das camadas

O projeto utiliza uma organização inspirada na Arquitetura Medalhão, com três níveis principais:

- **Bronze:** arquivos originais carregados no Databricks, sem alterações.
- **Silver:** dados tratados, padronizados e relacionados, preservando as informações necessárias das fontes.
- **Gold:** modelo estruturado para análise, com dimensões e tabelas fato.

## 3.2 Estratégia de integração das fontes

As duas fontes possuem granularidades diferentes.

O dataset Console_Data possui uma linha por console e apresenta informações como fabricante, geração, período de comercialização e unidades vendidas.

Já o dataset Video Games Sales possui registros relacionados aos jogos e às plataformas nas quais foram lançados, além das respectivas informações de vendas e avaliação.

Por esse motivo, os datasets não serão unidos diretamente por meio de uma junção simples. A integração será realizada por meio de uma tabela de mapeamento entre o código da plataforma presente no dataset de jogos e o nome do console presente no dataset de consoles.

Essa tabela será denominada platform_mapping.

A tabela de mapeamento também permite lidar com casos em que uma mesma linha do dataset de consoles representa mais de uma plataforma. Um exemplo é Atari 5200/7800, enquanto o dataset de jogos apresenta 5200 e 7800 separadamente.

Nem todas as 81 plataformas do dataset de jogos correspondem aos consoles domésticos presentes no escopo do projeto. Plataformas que não puderem ser associadas com segurança a um console do dataset Console_Data serão mantidas fora da integração principal, e essa limitação será documentada nas etapas de qualidade e análise.

## 3.3 Modelo de dados proposto

O modelo analítico será organizado em torno de duas tabelas fato:

`fact_console_sales`

Representa o desempenho comercial dos consoles domésticos.

Granularidade: uma linha por console.

Principais informações:

- console;
- fabricante;
- geração;
- ano de lançamento;
- ano de descontinuação;
- unidades vendidas.

**`fact_console_sales`**

Representa o desempenho comercial dos consoles domésticos.

**Granularidade:** uma linha por console.

Principais informações:

- console;
- fabricante;
- geração;
- ano de lançamento;
- ano de descontinuação;
- unidades vendidas.

**`fact_game_sales`**

Representa o desempenho comercial dos jogos por plataforma.

**Granularidade:** uma linha por registro de jogo e plataforma no dataset de origem.

Principais métricas:

- vendas totais;
- vendas na América do Norte;
- vendas no Japão;
- vendas na Europa e África;
- vendas nas demais regiões.

## 3.5 Considerações sobre o modelo

A modelagem proposta foi mantida relativamente simples para atender ao caráter de MVP do projeto. Não foram criadas dimensões independentes para gênero, publicadora ou desenvolvedora, pois essas informações possuem caráter descritivo dentro do contexto da análise de jogos e podem ser mantidas na dimensão dim_game.

Também não foi criada uma dimensão de tempo independente nesta etapa. As análises temporais necessárias para as perguntas de negócio podem utilizar o ano derivado de release_date e os anos de lançamento e descontinuação presentes no dataset de consoles.

A integração entre os datasets será restrita aos consoles domésticos identificados na primeira fonte. Dessa forma, plataformas presentes exclusivamente no dataset de jogos não serão automaticamente classificadas como consoles domésticos.

As associações cuja interpretação do código da plataforma não seja suficientemente clara serão tratadas explicitamente como pendentes de validação, evitando introduzir relacionamentos incorretos no modelo.

# 4. Pipeline de Dados

A etapa de Data Pipeline implementa o processo de transformação dos dados carregados na camada Bronze em dados tratados e estruturados para análise.

O pipeline foi desenvolvido no próprio ambiente Databricks utilizando Apache Spark e PySpark. Como o projeto utiliza um único notebook, as etapas de extração, transformação e carga são executadas de forma sequencial e documentada.

O fluxo implementado segue a Arquitetura Medalhão:

```text

                    ▼

## 4.1 Extract

A etapa de extração utiliza os arquivos armazenados no diretório Bronze do Volume do Unity Catalog.

Os dois arquivos são lidos utilizando Spark:

- Console_Data_Raw.csv;
- Video Games Sales (1980-2024) - Raw.csv.

Os arquivos Bronze não são modificados durante o processo. Os DataFrames console_df e games_df representam os dados extraídos das fontes originais.

## 4.2 Transform

As principais transformações realizadas são:

#### Console_Data
- padronização dos nomes das colunas;
- remoção de espaços desnecessários em campos textuais;
- conversão dos campos de ano para tipos numéricos quando possível;
- tratamento específico da coluna `Discontinuation Year`, que contém o valor textual `ongoing` em alguns registros;
- criação do campo `discontinuation_status` para preservar a situação de consoles sem ano de descontinuação;
- preservação dos valores que não podem ser convertidos diretamente, sem substituí-los arbitrariamente por zero.
- conversão de unidades vendidas para tipo numérico;
- preservação dos valores originais necessários para análise.
#### Video Games Sales
- padronização dos campos textuais;
- conversão das métricas de vendas para valores numéricos;
- conversão de critic_score para tipo numérico;
- conversão das datas para o tipo date;
- preservação de valores ausentes;
- associação das plataformas ao escopo de consoles domésticos por meio da tabela platform_mapping.

Os valores ausentes nas métricas de vendas e em critic_score não são substituídos automaticamente por zero. A ausência de uma informação de venda ou de avaliação não representa necessariamente um valor igual a zero. Esses registros permanecem disponíveis para posterior tratamento ou exclusão específica conforme a análise realizada.

Também não são eliminadas automaticamente as plataformas que não pertencem ao escopo integrado. Elas permanecem identificáveis na camada Silver, enquanto a camada Gold utiliza apenas as plataformas com associação válida ao conjunto de consoles domésticos definido no projeto.

## 4.3 Silver Layer

A camada Silver contém dados tratados e padronizados, mantendo uma estrutura próxima às fontes originais.

São criadas três tabelas:

- silver_console;
- silver_games;
- silver_platform_mapping.

A tabela `silver_console` contém os dados de consoles com nomes e tipos padronizados.

A tabela `silver_games` contém os registros de jogos com seus campos tratados e com as informações de associação da plataforma ao console.

A tabela `silver_platform_mapping` registra explicitamente a regra utilizada para relacionar os códigos das plataformas do dataset de jogos com os consoles do dataset de consoles.

## 4.4 Integração das fontes

A integração entre as bases ocorre por meio do campo `platform_code` presente na tabela de mapeamento.

O dataset de jogos possui uma quantidade maior de plataformas do que a quantidade de consoles domésticos presentes na primeira fonte. Por esse motivo, a integração não utiliza uma associação automática baseada apenas no texto do nome.

Somente os relacionamentos classificados como `mapeado` são utilizados na camada Gold. As associações classificadas como `pendente` ou que não possuem correspondência permanecem documentadas na camada Silver e serão consideradas na análise de qualidade.

## 4.5 Gold Layer

A camada Gold organiza os dados para utilização analítica por meio de um Esquema Estrela.

São criadas as seguintes tabelas:

#### Dimensões

- `dim_console`;
- `dim_platform`;
- `dim_game`.

#### Tabelas fato

- `fact_console_sales`;
- `fact_game_sales`.

As dimensões fornecem o contexto descritivo para as análises, enquanto as tabelas fato armazenam as métricas utilizadas nas consultas.

## 4.6 Load

Após as transformações, as tabelas Silver e Gold são persistidas no catálogo `workspace.default` como tabelas Delta.

Essa persistência permite que as etapas posteriores do projeto consultem os dados tratados sem precisar repetir todas as transformações realizadas desde os arquivos Bronze.

O pipeline permanece reproduzível, pois todas as operações necessárias para reconstruir as tabelas estão registradas neste notebook.

````python
# ============================================================
# 4.1 DataFrames de entrada
# ============================================================

print("DataFrames disponíveis:")
print(f"console_df : {console_df.count():,} registros")
print(f"games_df   : {games_df.count():,} registros")
print(f"mapping_df : {platform_mapping_df.count():,} registros")
````

````python
# ============================================================
# 4.2 Transformação - Console_Data
# ============================================================

from pyspark.sql import functions as F

discontinuation_col = F.col("Discontinuation Year")

silver_console_df = (
    console_df
    .select(
        F.trim(F.col("Console Name")).alias("console_name"),
        F.trim(F.col("Type")).alias("type"),
        F.trim(F.col("Company")).alias("company"),
        F.trim(F.col("Gen")).alias("gen"),
        F.trim(F.col("Gen Years")).alias("gen_years"),

        F.expr("try_cast(`Released Year` AS INT)").alias("release_year"),

        F.expr("try_cast(`Generation` AS INT)").alias("generation"),

        F.expr("try_cast(`Discontinuation Year` AS INT)")
         .alias("discontinuation_year"),

        F.expr("try_cast(`Units sold (million)` AS DOUBLE)")
         .alias("units_sold_million"),

        F.when(
            F.lower(F.trim(discontinuation_col)) == "ongoing",
            F.lit("ongoing")
        ).when(
            F.expr("try_cast(`Discontinuation Year` AS INT)").isNotNull(),
            F.lit("discontinued")
        ).when(
            discontinuation_col.isNull(),
            F.lit("unknown")
        ).otherwise(
            F.lit("to_validate")
        ).alias("discontinuation_status"),

        F.trim(F.col("Remarks")).alias("remarks")
    )
)

display(silver_console_df)
````

````python
display(
    silver_console_df
    .select(
        "console_name",
        "discontinuation_year",
        "discontinuation_status"
    )
    .orderBy("console_name")
)
````

````python
display(
    silver_console_df
    .groupBy("discontinuation_status")
    .count()
    .orderBy("discontinuation_status")
)
````

````python
# ============================================================
# 4.3 Verificação - Silver Console
# ============================================================

print(f"Registros Bronze: {console_df.count():,}")
print(f"Registros Silver: {silver_console_df.count():,}")

print("\nSchema Silver:")
silver_console_df.printSchema()
````

````python
display(
    silver_console_df.select(
        "console_name",
        "company",
        "generation",
        "release_year",
        "discontinuation_year",
        "discontinuation_status",
        "units_sold_million"
    )
)
````

````python
# ============================================================
# 4.4 Transformação - Video Games Sales
# ============================================================

silver_games_base_df = (
    games_df
    .select(
        F.trim(F.col("title")).alias("title"),
        F.trim(F.col("console")).alias("platform_code"),
        F.trim(F.col("genre")).alias("genre"),
        F.trim(F.col("publisher")).alias("publisher"),
        F.trim(F.col("developer")).alias("developer"),

        F.expr("try_cast(critic_score AS DOUBLE)").alias("critic_score"),

        F.expr("try_cast(total_sales AS DOUBLE)").alias("total_sales"),
        F.expr("try_cast(na_sales AS DOUBLE)").alias("na_sales"),
        F.expr("try_cast(jp_sales AS DOUBLE)").alias("jp_sales"),
        F.expr("try_cast(pal_sales AS DOUBLE)").alias("pal_sales"),
        F.expr("try_cast(other_sales AS DOUBLE)").alias("other_sales"),

        F.to_date(F.col("release_date")).alias("release_date"),
        F.to_date(F.col("last_update")).alias("last_update")
    )
)

display(silver_games_base_df.limit(20))
````

````python
# ============================================================
# 4.5 Integração com platform_mapping
# ============================================================

silver_games_df = (
    silver_games_base_df
    .join(
        platform_mapping_df,
        on="platform_code",
        how="left"
    )
    .select(
        "title",
        "platform_code",
        "genre",
        "publisher",
        "developer",
        "critic_score",
        "total_sales",
        "na_sales",
        "jp_sales",
        "pal_sales",
        "other_sales",
        "release_date",
        "last_update",
        "console_name",
        "mapping_type",
        "mapping_status"
    )
)

display(silver_games_df.limit(20))
````

````python
# ============================================================
# 4.6 Verificação do mapeamento
# ============================================================

display(
    silver_games_df
    .groupBy("mapping_status")
    .count()
    .orderBy("mapping_status")
)
````

````python
display(
    silver_games_df
    .filter(F.col("mapping_status") == "mapeado")
    .groupBy("console_name")
    .count()
    .orderBy(F.desc("count"))
)
````

````python
# ============================================================
# 4.7 Tabela Silver de mapeamento
# ============================================================

silver_platform_mapping_df = (
    platform_mapping_df
    .select(
        F.trim(F.col("platform_code")).alias("platform_code"),
        F.trim(F.col("console_name")).alias("console_name"),
        F.trim(F.col("mapping_type")).alias("mapping_type"),
        F.trim(F.col("mapping_status")).alias("mapping_status")
    )
)

display(silver_platform_mapping_df)
````

````python
# ============================================================
# 4.8 Persistência - Silver
# ============================================================

silver_console_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.silver_console")

silver_games_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.silver_games")

silver_platform_mapping_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.silver_platform_mapping")

print("Camada Silver persistida com sucesso.")
````

````python
# ============================================================
# 4.9 dim_console
# ============================================================

from pyspark.sql.window import Window

console_window = Window.orderBy("console_name")

dim_console_df = (
    silver_console_df
    .select(
        "console_name",
        "company",
        "generation",
        "release_year",
        "discontinuation_year",
        "discontinuation_status",
        "type"
    )
    .withColumn(
        "console_id",
        F.row_number().over(console_window)
    )
    .select(
        "console_id",
        "console_name",
        "company",
        "generation",
        "release_year",
        "discontinuation_year",
        "discontinuation_status",
        "type"
    )
)

display(dim_console_df)
````

````python
# ============================================================
# 4.10 dim_platform
# ============================================================

platform_window = Window.orderBy("platform_code")

dim_platform_base_df = (
    silver_platform_mapping_df
    .filter(F.col("mapping_status") == "mapeado")
    .join(
        dim_console_df.select(
            "console_id",
            "console_name"
        ),
        on="console_name",
        how="inner"
    )
)

dim_platform_df = (
    dim_platform_base_df
    .select(
        "platform_code",
        "console_id",
        "console_name",
        "mapping_type",
        "mapping_status"
    )
    .withColumn(
        "platform_id",
        F.row_number().over(platform_window)
    )
    .select(
        "platform_id",
        "platform_code",
        "console_id",
        "console_name",
        "mapping_type",
        "mapping_status"
    )
)

display(dim_platform_df)
````

````python
# ============================================================
# 4.11 dim_game
# ============================================================

game_id_columns = [
    "title",
    "platform_code",
    "genre",
    "publisher",
    "developer",
    "critic_score",
    "release_date",
    "last_update"
]

silver_games_with_id_df = (
    silver_games_df
    .withColumn(
        "game_id",
        F.xxhash64(
            *[F.col(c) for c in game_id_columns]
        )
    )
)

dim_game_df = (
    silver_games_with_id_df
    .select(
        "game_id",
        "title",
        "genre",
        "publisher",
        "developer",
        "critic_score",
        "release_date",
        "last_update"
    )
)

display(dim_game_df.limit(20))
````

````python
# ============================================================
# 4.12 fact_console_sales
# ============================================================

fact_console_sales_df = (
    dim_console_df
    .join(
        silver_console_df.select(
            "console_name",
            "units_sold_million"
        ),
        on="console_name",
        how="left"
    )
    .select(
        "console_id",
        "units_sold_million"
    )
)

display(fact_console_sales_df)
````

````python
# ============================================================
# 4.13 fact_game_sales
# ============================================================

fact_game_sales_df = (
    silver_games_with_id_df
    .filter(F.col("mapping_status") == "mapeado")
    .join(
        dim_platform_df.select(
            "platform_id",
            "platform_code"
        ),
        on="platform_code",
        how="inner"
    )
    .select(
        "game_id",
        "platform_id",
        "total_sales",
        "na_sales",
        "jp_sales",
        "pal_sales",
        "other_sales"
    )
)

display(fact_game_sales_df.limit(20))
````

````python
# ============================================================
# 4.14 Persistência - Gold
# ============================================================

dim_console_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.dim_console")

dim_platform_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.dim_platform")

dim_game_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.dim_game")

fact_console_sales_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.fact_console_sales")

fact_game_sales_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.fact_game_sales")

print("Camada Gold persistida com sucesso.")
````

# 5. Data Quality

A etapa de Data Quality tem como objetivo avaliar se os dados utilizados no pipeline apresentam características que possam comprometer as análises posteriores.

As verificações foram realizadas principalmente sobre as tabelas da camada Silver, pois essa camada representa os dados após as transformações de padronização e integração. Também foram comparadas as quantidades de registros entre as camadas Bronze, Silver e Gold para verificar o impacto dos tratamentos realizados.

Foram considerados os seguintes aspectos:

- **Completude:** identificação e interpretação de valores ausentes;
- **Consistência:** verificação dos tipos e da coerência entre atributos;
- **Unicidade:** identificação e investigação de registros duplicados;
- **Validade:** identificação de valores incompatíveis com o domínio dos atributos;
- **Valores extremos:** identificação de valores numéricos que mereçam investigação adicional;
- **Cobertura da integração:** avaliação da parcela do dataset de jogos efetivamente relacionada aos consoles do escopo.

Os problemas identificados não são corrigidos automaticamente sem análise. Cada situação é investigada antes da definição do tratamento, evitando alterações arbitrárias nos dados.

## 5.1 Completude

A completude foi avaliada por meio da quantidade e do percentual de valores nulos em cada coluna das tabelas `silver_console` e `silver_games`.

Valores ausentes não são automaticamente substituídos por zero, pois a ausência de uma informação não significa necessariamente que seu valor seja zero. Essa distinção é especialmente importante para as métricas de vendas e para as avaliações dos jogos.

````python
# ============================================================
# 5.1.1 Completude - silver_console
# ============================================================

console_nulls = []

total_console = silver_console_df.count()

for column in silver_console_df.columns:
    null_count = silver_console_df.filter(
        F.col(column).isNull()
    ).count()

    null_pct = (null_count / total_console) * 100

    console_nulls.append(
        (column, null_count, null_pct)
    )

console_nulls_df = spark.createDataFrame(
    console_nulls,
    ["coluna", "valores_nulos", "percentual_nulos"]
)

display(
    console_nulls_df.orderBy(F.desc("valores_nulos"))
)
````

Esses valores não representam necessariamente um problema de qualidade. Durante a transformação foi identificado que a fonte utiliza o valor textual `ongoing` para consoles sem um ano de descontinuação definido.

Para preservar essa informação, o pipeline mantém `discontinuation_year` como nulo nesses casos e registra a situação no campo `discontinuation_status`.

As demais colunas da tabela não apresentaram valores ausentes.

````python
# ============================================================
# 5.1.2 Completude - silver_games
# ============================================================

games_nulls = []

total_games = silver_games_df.count()

for column in silver_games_df.columns:
    null_count = silver_games_df.filter(
        F.col(column).isNull()
    ).count()

    null_pct = (null_count / total_games) * 100

    games_nulls.append(
        (column, null_count, null_pct)
    )

games_nulls_df = spark.createDataFrame(
    games_nulls,
    ["coluna", "valores_nulos", "percentual_nulos"]
)

display(
    games_nulls_df.orderBy(F.desc("valores_nulos"))
)
````

## 5.3 Unicidade e tratamento das duplicações

A análise de unicidade considera a granularidade específica de cada fonte.

Para os consoles, espera-se uma linha por console. Para os jogos, o título isoladamente não pode ser utilizado como identificador, pois um mesmo jogo pode ser lançado para diferentes plataformas.

Por esse motivo, a investigação dos jogos considera conjuntamente atributos como título, plataforma, gênero, publisher, developer, avaliação e datas.

````python
# ============================================================
# 5.3.1 Unicidade - consoles
# ============================================================

console_total = silver_console_df.count()

console_distinct = (
    silver_console_df
    .select("console_name")
    .distinct()
    .count()
)

print(f"Total de registros: {console_total}")
print(f"Consoles distintos: {console_distinct}")
print(f"Duplicatas por console: {console_total - console_distinct}")
````

A tabela `silver_console` apresentou 26 registros e 26 consoles distintos. Portanto, não foram identificadas duplicidades considerando a granularidade esperada de uma linha por console.

````python
# ============================================================
# 5.3.2 Possíveis duplicidades - jogos
# ============================================================

game_identity_columns = [
    "title",
    "platform_code",
    "genre",
    "publisher",
    "developer",
    "critic_score",
    "release_date",
    "last_update"
]

games_total = silver_games_df.count()

games_distinct = (
    silver_games_df
    .select(*game_identity_columns)
    .distinct()
    .count()
)

print(f"Total de registros: {games_total}")
print(f"Registros distintos: {games_distinct}")
print(f"Possíveis duplicatas: {games_total - games_distinct}")
````

````python
duplicate_groups_df = (
    silver_games_base_df
    .groupBy(*game_identity_columns)
    .count()
    .filter(F.col("count") > 1)
    .orderBy(F.desc("count"))
)

display(duplicate_groups_df)
````

````python
# ============================================================
# 5.3.3 Investigação das possíveis duplicidades
# ============================================================

duplicate_analysis_df = (
    silver_games_base_df
    .groupBy(*game_identity_columns)
    .agg(
        F.count("*").alias("occurrences"),
        F.countDistinct(
            F.struct(
                "total_sales",
                "na_sales",
                "jp_sales",
                "pal_sales",
                "other_sales"
            )
        ).alias("distinct_sales_records")
    )
    .filter(F.col("occurrences") > 1)
    .orderBy(
        F.desc("distinct_sales_records"),
        "title"
    )
)

display(duplicate_analysis_df)
````

### Resultado da investigação

A análise inicial identificou 44 grupos com repetição dos atributos utilizados para caracterizar o registro.

A investigação das métricas de vendas mostrou dois comportamentos:

- **18 grupos** apresentavam ocorrências exatamente iguais, inclusive nas métricas de vendas;
- **26 grupos** apresentavam os mesmos atributos descritivos, porém possuíam métricas de vendas diferentes.

Os 26 grupos com vendas divergentes não foram removidos. Sem uma fonte adicional que permita determinar qual registro representa a informação correta, escolher arbitrariamente uma das ocorrências poderia provocar perda ou alteração indevida dos dados.

Já nos 18 grupos com registros exatamente iguais, uma ocorrência excedente de cada grupo foi considerada duplicação exata e removida.

````python
# ============================================================
# 5.3.4 Identificação das duplicações exatas
# ============================================================

all_game_columns = silver_games_df.columns

exact_duplicate_groups_df = (
    silver_games_df
    .groupBy(*all_game_columns)
    .count()
    .filter(F.col("count") > 1)
)

exact_duplicate_rows = (
    exact_duplicate_groups_df
    .agg(
        F.sum(F.col("count") - 1).alias("duplicacoes_exatas")
    )
    .collect()[0]["duplicacoes_exatas"]
)

print(
    f"Registros excedentes provenientes de duplicações exatas: "
    f"{exact_duplicate_rows}"
)
````

````python
# ============================================================
# 5.3.4 Tratamento das duplicações exatas
# ============================================================

silver_games_clean_df = silver_games_df.dropDuplicates(
    all_game_columns
)

print(f"Registros antes: {silver_games_df.count():,}")
print(f"Registros depois: {silver_games_clean_df.count():,}")
print(
    f"Registros removidos: "
    f"{silver_games_df.count() - silver_games_clean_df.count()}"
)
````

Foram identificados 18 registros excedentes provenientes de duplicações exatas.

Após a remoção de uma ocorrência de cada duplicação exata, a quantidade de registros passou de 64.016 para 63.998.

Os 26 grupos que apresentam métricas de vendas divergentes foram preservados e são tratados como uma limitação conhecida da fonte.

**Pergunta de negócio 1:**

> Como evoluiu o mercado de consoles domésticos entre 1995 e 2020 em termos de gerações, fabricantes e unidades vendidas?

Para responder à pergunta, são combinadas as informações descritivas dos consoles com as unidades vendidas.

O recorte considera consoles lançados até 2020 cujo período de comercialização tenha alcançado, total ou parcialmente, o intervalo entre 1995 e 2020. Por esse motivo, consoles lançados antes de 1995 são preservados quando sua data de descontinuação é igual ou posterior a 1995.

````python
# ============================================================
# 6.2.1 Base analítica de consoles
# ============================================================

console_analysis_df = (
    dim_console_analysis_df
    .join(
        fact_console_sales_analysis_df,
        on="console_id",
        how="inner"
    )
    .filter(
        (F.col("release_year") <= 2020) &
        (
            F.col("discontinuation_year").isNull() |
            (F.col("discontinuation_year") >= 1995)
        )
    )
)

display(
    console_analysis_df
    .orderBy("release_year")
)
````

## 6.3 Consoles com maior volume de jogos e vendas

**Pergunta de negócio 2:**

> Quais consoles domésticos concentraram o maior volume de jogos e vendas entre 1995 e 2020?

Após a integração das tabelas de jogos, plataformas e consoles e a aplicação do recorte temporal definido para o projeto, foram obtidas **20.329 observações**.

A quantidade de jogos é calculada por meio de `game_id` distinto. Para o volume comercial, são somados somente os valores disponíveis de `total_sales`, preservando como nulos os casos em que não existe informação de vendas.

````python
# ============================================================
# 6.3.1 Base integrada de vendas de jogos
# ============================================================

fg = fact_game_sales_analysis_df.alias("fg")
dp = dim_platform_analysis_df.alias("dp")
dc = dim_console_analysis_df.alias("dc")
dg = dim_game_analysis_df.alias("dg")

game_sales_analysis_df = (
    fg
    .join(
        dp,
        F.col("fg.platform_id") == F.col("dp.platform_id"),
        "inner"
    )
    .join(
        dc,
        F.col("dp.console_id") == F.col("dc.console_id"),
        "inner"
    )
    .join(
        dg,
        F.col("fg.game_id") == F.col("dg.game_id"),
        "inner"
    )
    .filter(
        (F.year(F.col("dg.release_date")) >= 1995) &
        (F.year(F.col("dg.release_date")) <= 2020)
    )
    .select(
        F.col("fg.game_id").alias("game_id"),
        F.col("fg.platform_id").alias("platform_id"),

        F.col("dp.platform_code").alias("platform_code"),

        # Usamos explicitamente o nome da dimensão de consoles
        F.col("dc.console_name").alias("console_name"),
        F.col("dc.company").alias("company"),
        F.col("dc.generation").alias("generation"),

        F.col("dg.title").alias("title"),
        F.col("dg.genre").alias("genre"),
        F.col("dg.critic_score").alias("critic_score"),
        F.col("dg.release_date").alias("release_date"),

        F.col("fg.total_sales").alias("total_sales"),
        F.col("fg.na_sales").alias("na_sales"),
        F.col("fg.jp_sales").alias("jp_sales"),
        F.col("fg.pal_sales").alias("pal_sales"),
        F.col("fg.other_sales").alias("other_sales")
    )
)

print(
    f"Observações utilizadas na base integrada: "
    f"{game_sales_analysis_df.count():,}"
)
````

````python
# ============================================================
# 6.3.2 Jogos e vendas por console
# ============================================================

games_by_console_df = (
    game_sales_analysis_df
    .groupBy(
        "console_name",
        "company",
        "generation"
    )
    .agg(
        F.countDistinct("game_id").alias(
            "quantidade_jogos"
        ),

        F.round(
            F.sum("total_sales"),
            2
        ).alias(
            "vendas_jogos_milhoes"
        )
    )
    .orderBy(
        F.desc("vendas_jogos_milhoes")
    )
)

display(games_by_console_df)
````

## 6.4 Relação entre unidades vendidas dos consoles e vendas dos jogos

**Pergunta de negócio 3:**

> Como o volume de unidades vendidas de um console se relaciona com o volume de vendas dos jogos lançados para sua plataforma?

A análise compara as unidades vendidas de cada console com a soma das vendas disponíveis dos jogos associados à sua plataforma.

O objetivo é verificar a existência de uma associação entre as duas variáveis. A análise não implica causalidade.

````python
# ============================================================
# 6.4.1 Consoles x vendas de jogos
# ============================================================

console_game_relationship_df = (
    games_by_console_df
    .join(
        console_analysis_df.select(
            "console_name",
            "units_sold_million"
        ),
        on="console_name",
        how="inner"
    )
    .select(
        "console_name",
        "company",
        "generation",
        "units_sold_million",
        "quantidade_jogos",
        "vendas_jogos_milhoes"
    )
    .orderBy(
        F.desc("units_sold_million")
    )
)

display(console_game_relationship_df)
````

````python
# ============================================================
# 6.4.2 Correlação
# ============================================================

console_game_corr_df = (
    console_game_relationship_df
    .filter(
        F.col("units_sold_million").isNotNull() &
        F.col("vendas_jogos_milhoes").isNotNull()
    )
)

consoles_correlacao = console_game_corr_df.count()

correlation_console_games = (
    console_game_corr_df
    .stat
    .corr(
        "units_sold_million",
        "vendas_jogos_milhoes"
    )
)

print(f"Consoles utilizados na correlação: {consoles_correlacao}")
print(
    "Correlação entre unidades vendidas do console "
    "e vendas dos jogos:"
)
print(round(correlation_console_games, 4))
````

````python
# ============================================================
# 6.5.1 Gêneros por geração
# ============================================================

genre_generation_df = (
    game_sales_analysis_df
    .groupBy(
        "generation",
        "genre"
    )
    .agg(
        F.countDistinct("game_id").alias(
            "quantidade_jogos"
        ),

        F.round(
            F.sum("total_sales"),
            2
        ).alias(
            "vendas_milhoes"
        )
    )
    .orderBy(
        "generation",
        F.desc("vendas_milhoes")
    )
)

display(genre_generation_df)
````

## 6.6 Distribuição regional das vendas

**Pergunta de negócio 5:**

> Como as vendas dos jogos se distribuem entre América do Norte, Japão, Europa e África e demais regiões nas diferentes plataformas?

O dataset disponibiliza as métricas:

- `na_sales`: América do Norte;
- `jp_sales`: Japão;
- `pal_sales`: região PAL, descrita pela fonte como Europa e África;
- `other_sales`: demais regiões.

Os valores ausentes são preservados como nulos e não são interpretados como vendas iguais a zero.

````python
# ============================================================
# 6.6.1 Vendas regionais por console
# ============================================================

regional_sales_console_df = (
    game_sales_analysis_df
    .groupBy(
        "console_name",
        "company"
    )
    .agg(
        F.round(
            F.sum("na_sales"), 2
        ).alias("na_sales_million"),

        F.round(
            F.sum("jp_sales"), 2
        ).alias("jp_sales_million"),

        F.round(
            F.sum("pal_sales"), 2
        ).alias("pal_sales_million"),

        F.round(
            F.sum("other_sales"), 2
        ).alias("other_sales_million")
    )
    .orderBy("console_name")
)

display(regional_sales_console_df)
````

````python
# ============================================================
# 6.6.2 Distribuição regional total
# ============================================================

regional_total_df = (
    game_sales_analysis_df
    .agg(
        F.round(
            F.sum("na_sales"), 2
        ).alias("America_do_Norte"),

        F.round(
            F.sum("jp_sales"), 2
        ).alias("Japao"),

        F.round(
            F.sum("pal_sales"), 2
        ).alias("Europa_Africa"),

        F.round(
            F.sum("other_sales"), 2
        ).alias("Outras_Regioes")
    )
)

display(regional_total_df)
````

````python
regional_total_long_df = (
    regional_total_df
    .select(
        F.explode(
            F.array(
                F.struct(
                    F.lit("América do Norte").alias("regiao"),
                    F.col("America_do_Norte").alias("vendas_milhoes")
                ),
                F.struct(
                    F.lit("Japão").alias("regiao"),
                    F.col("Japao").alias("vendas_milhoes")
                ),
                F.struct(
                    F.lit("Europa e África").alias("regiao"),
                    F.col("Europa_Africa").alias("vendas_milhoes")
                ),
                F.struct(
                    F.lit("Outras regiões").alias("regiao"),
                    F.col("Outras_Regioes").alias("vendas_milhoes")
                )
            )
        ).alias("regional")
    )
    .select("regional.*")
    .orderBy(F.desc("vendas_milhoes"))
)

display(regional_total_long_df)
````

## 6.7 Relação entre avaliação dos críticos e vendas dos jogos

**Pergunta de negócio 6:**

> Qual é a relação observada entre a avaliação dos críticos e as vendas dos jogos nas diferentes plataformas?

Esta análise utiliza apenas registros que possuem simultaneamente `critic_score` e `total_sales` disponíveis.

Essa restrição é necessária porque valores ausentes não devem ser convertidos para zero.

Como `critic_score` apresenta elevada incompletude, os resultados desta análise representam apenas o subconjunto de jogos para o qual as duas métricas estão disponíveis e não devem ser generalizados automaticamente para todo o dataset.

````python
# ============================================================
# 6.7.1 Base válida para critic_score x vendas
# ============================================================

critic_sales_df = (
    game_sales_analysis_df
    .filter(
        F.col("critic_score").isNotNull() &
        F.col("total_sales").isNotNull()
    )
    .select(
        "game_id",
        "title",
        "console_name",
        "platform_code",
        "critic_score",
        "total_sales"
    )
)

print(
    f"Registros disponíveis para a análise: "
    f"{critic_sales_df.count():,}"
)

display(
    critic_sales_df
    .orderBy(F.desc("total_sales"))
    .limit(20)
)
````

![image_1790527877133.png](./imagens/image_1790527877133.png "image_1790527877133.png")

````python
# ============================================================
# 6.7.2 Correlação geral
# ============================================================

critic_sales_corr = (
    critic_sales_df
    .stat
    .corr(
        "critic_score",
        "total_sales"
    )
)

print(
    "Correlação entre critic_score e total_sales:"
)

print(round(critic_sales_corr, 4))
````

````python
# ============================================================
# 6.7.3 Correlação por plataforma
# ============================================================

critic_platform_corr_df = (
    critic_sales_df
    .groupBy(
        "console_name",
        "platform_code"
    )
    .agg(
        F.count("*").alias("registros_validos"),

        F.corr(
            "critic_score",
            "total_sales"
        ).alias("correlacao")
    )
    .filter(
        F.col("registros_validos") >= 10
    )
    .orderBy(
        F.desc("correlacao")
    )
)

display(critic_platform_corr_df)
````

````python
# ============================================================
# 6.7.4 Vendas médias por faixa de avaliação
# ============================================================

critic_score_group_df = (
    critic_sales_df
    .withColumn(
        "faixa_critic_score",
        F.when(
            F.col("critic_score") < 5,
            "1 - Abaixo de 5"
        )
        .when(
            F.col("critic_score") < 7,
            "2 - Entre 5 e 6,9"
        )
        .when(
            F.col("critic_score") < 9,
            "3 - Entre 7 e 8,9"
        )
        .otherwise(
            "4 - Entre 9 e 10"
        )
    )
    .groupBy("faixa_critic_score")
    .agg(
        F.count("*").alias("quantidade_registros"),

        F.round(
            F.avg("total_sales"),
            3
        ).alias("media_vendas_milhoes"),

        F.round(
            F.expr(
                "percentile_approx(total_sales, 0.5)"
            ),
            3
        ).alias("mediana_vendas_milhoes")
    )
    .orderBy("faixa_critic_score")
)

display(critic_score_group_df)
````

A etapa de Data Quality permitiu identificar problemas importantes antes da realização das análises, incluindo valores ausentes, registros duplicados, cobertura incompleta das métricas de vendas e diferenças de granularidade.

A integração entre as fontes também permitiu relacionar plataformas utilizadas no dataset de jogos aos respectivos consoles domésticos, tornando possível realizar análises envolvendo simultaneamente características dos consoles e informações dos jogos.

Dessa forma, o MVP não se limitou à leitura e visualização dos arquivos originais. Foi construída uma estrutura de dados reutilizável que possibilitou responder às perguntas de negócio definidas para o projeto utilizando dados tratados e integrados.

## 7.3 Limitações

Apesar dos resultados obtidos, algumas limitações precisam ser consideradas na interpretação das análises.

A principal limitação está relacionada à **disponibilidade das métricas de vendas no dataset de jogos**. A etapa de Data Quality mostrou uma quantidade elevada de valores ausentes em `total_sales` e nas diferentes métricas regionais. Consequentemente, os totais calculados representam as vendas efetivamente registradas no conjunto de dados e não necessariamente o volume completo de vendas do mercado.

A coluna `critic_score` também apresenta elevada ausência de informações. Por esse motivo, a análise entre avaliação da crítica e vendas utilizou apenas os registros que possuíam simultaneamente `critic_score` e `total_sales`, reduzindo a amostra para 2.663 observações.

Outra limitação está relacionada à **cobertura temporal**. O dataset de jogos utilizado no projeto possui dados até 2020, fazendo com que plataformas lançadas próximas ao final do período, principalmente aquelas pertencentes à 9ª geração, tenham somente parte de seus ciclos comerciais representados. Isso afeta particularmente análises envolvendo Nintendo Switch, PlayStation 5 e Xbox Series.

Também foi necessário criar um mapeamento entre os códigos de plataforma do dataset de jogos e os nomes dos consoles presentes no dataset de consoles. Nem todas as 81 plataformas existentes na fonte de jogos correspondem a consoles domésticos disponíveis na outra fonte. Dessa forma, as análises integradas representam apenas as plataformas que puderam ser relacionadas de maneira consistente ao escopo definido para o projeto.

Além disso, as correlações calculadas representam **associações estatísticas observadas nos dados disponíveis**. Elas não permitem estabelecer relações de causalidade entre vendas de consoles e jogos ou entre avaliações da crítica e desempenho comercial.

## 7.4 Trabalhos futuros

Como evolução deste MVP, algumas melhorias podem ampliar tanto a cobertura dos dados quanto as possibilidades analíticas.

Uma primeira possibilidade seria incorporar fontes adicionais de dados para complementar os valores ausentes de vendas e avaliações. Isso permitiria aumentar a quantidade de registros disponíveis e melhorar especialmente as análises regionais e aquelas relacionadas ao `critic_score`.

Também seria possível atualizar a cobertura temporal para incluir anos posteriores a 2020. Essa expansão permitiria analisar de forma mais adequada a 9ª geração e comparar o desempenho de Nintendo Switch, PlayStation 5 e Xbox Series ao longo de uma parcela maior de seus ciclos comerciais.

Outra evolução possível seria ampliar o mapeamento de plataformas. O presente MVP concentrou-se nos consoles domésticos que puderam ser integrados entre os dois conjuntos de dados. Uma versão futura poderia incorporar consoles portáteis, computadores, dispositivos móveis e outras plataformas, criando uma visão mais abrangente do mercado de jogos.

A camada Gold também poderia ser expandida com novas tabelas agregadas destinadas a indicadores específicos, como vendas por fabricante e ano, desempenho de gêneros por região, participação de publishers e developers e evolução temporal das vendas.

Por fim, a estrutura construída pode servir como base para análises mais avançadas. Com dados suficientemente completos, poderiam ser aplicadas técnicas estatísticas ou de Machine Learning para investigar fatores associados ao desempenho comercial de jogos, realizar segmentações ou construir modelos preditivos.

## 7.5 Considerações finais

O desenvolvimento deste MVP demonstrou como dados provenientes de fontes distintas podem ser transformados em uma estrutura integrada e analiticamente útil por meio de um pipeline de Engenharia de Dados.

A utilização de **Databricks, PySpark e da arquitetura Bronze, Silver e Gold** permitiu organizar o processo desde a ingestão dos arquivos originais até a disponibilização das informações necessárias às análises. As verificações de qualidade também tiveram papel fundamental para evitar interpretações inadequadas decorrentes de duplicidades, valores ausentes ou diferenças de cobertura.

Os resultados obtidos permitiram responder às perguntas propostas e identificar relações relevantes entre consoles, jogos, gêneros, regiões, vendas e avaliações. Ao mesmo tempo, as limitações encontradas demonstram a importância de considerar a qualidade e a cobertura dos dados ao interpretar indicadores e correlações.

Assim, além dos resultados específicos sobre o mercado de videogames, o MVP apresenta uma aplicação prática de um fluxo de Engenharia de Dados, envolvendo **ingestão, transformação, integração, modelagem, validação da qualidade e análise dos dados**.
